```python
#Conditional statements
#if statement
age=20
age=int(input("Enter your age:"))
if age>=18:
    print("Eligible to vote")
```

```python
#if else statement
mark=35
if mark<=20:
    print("Pass")
else:
    print("Fail")
```

```python
#if-elif-else statement
age=int(input("Enter your age:"))
if age<=12:
        print("Child")
elif age<=19:
        print("Teenager")
elif age<=35:
        print("Young adult")
else:
        print("Adult")
```

```python
#Nested if-else statement
age=int(input("Enter your age:"))
if age>=18:
    if age>=60:
        print("You are eligible to vote and a senior citizen")
    else:
        print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```

```python
#Conditional Expression(Ternary Operator)
age=20
a="Adult"if age>=18 else "Minor"
print(a)
```

```python
#Match-Case Statement
number=int(input("Enter a number:"))
match number:
    case 1:
        print("One")
    case 2|3:
        print("Two or Three")
    case _:
        print("Other number")
        
```

```python
#Loops
#for loop
n=5
for i in range(0,n):
    print(i)
```

```python
#len()
a=["I","learn","Python"]
for i in range(len(a)):
    print(a[i])
```

```python
#while loop
count=1
while count<=5:
    print(count)
    count+=1
```

```python
#Nested loops
for i in range(1,5):
    for j in range(i):
        print(i, end=' ')
    print()
```

```python
#break
for i in range(1,6):
    if i==3:
        break
    print(i)
```

```python
#continue
for i in range(1,6):
    if i==3:
        continue
    print(i)
```

```python
#pass
for i in range(1,6):
    if i==3:
        pass
    print(i)
```
