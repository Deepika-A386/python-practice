```python
      #Operators
#Arithmetic operators
a=30
b=2
print("Addition:", a+b)
print("Subtraction:", a-b)
print("Multiplication:", a*b)
print("Division:", a/b)
print("Floor Division:", a//b)
print("Modulus:", a%b)
print("Exponentiation:", a**b)
```

```python
#Comparison operators
a=23
b=13
print(a>b)
print(a<b)
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
```

```python
#Logical operators
a=True
b=False
print(a and b)   #logical and
print(a or b)    #logical or
print(not a)     #logical not
```

```python
#Bitwise operators 
a=10
b=3
print(a&b)   #
print(a|b)
print(~a)
print(a^b)
print(a>>2)
print(a<<2)
```

```python
#Assignment operators
a=12
b=a
print(b)
b+=a
print(b)
b-=a
print(b)
b*=a
print(b)
b<<=a
print(b)
```

```python
#Identity operators
a=20
b=80
c=a

print(a is not b)
print(a is c)
```

```python
#Incrementing & Decrementing 
x=2
x+=2
print(x)

y=20
y-=1
print(y)
```

```python
#operator associativity
print(200/20*20)
print(4-2+3)
print(4-(7-2))
print(2**4**1)
```

```python
import keyword         #brings a module
print(keyword.kwlist)  #list of keywords
```

```python
#Literals 
#Numeric literals
a=20
b=-20
c=0.20
d=-0.002
e=2+3j
f=-4j

print(a,b,c,d,e,f)
```

```python
#String literals
a='Hello'
b="Python"
c='''This is a multi-line string'''
d=r"C:\Users\Python"                #r-treat backslashes \ as normal characters

print(a)
print(b)
print(c)
print(d)
```

```python
#Boolean literals
a=True
b=False

print(a,b)
print(1==True)   
print(0==False)
print(True+2)
print(False+6)
```

```python
#Collection literals
rank=["First"]
colors=("Red")
student={"Ram":10}
nums={1,3}

print(rank)
print(colors)
print(student)
print(nums)
```

```python
#index
a="Deepika"
b=["Deepi","Ram","Indhu"]
t=(1,0)

print(a[3])
print(a[0])
print(b[2])
print(b[-1])
print(type(t))
```

```python
#Set data type
s1={"a","b","a","c"}
print(s1)
```

```python
#Dictionary data type
d={1:'one',2:'two',3:'three'}
print(d[1])
print(d.get(2))
```

```python
#Primitive data types
a=30
b=1.2
is_active=True
name="Deepi"

print(type(a))
print(type(b))
print(type(is_active))
print(type(name))
```

```python
#Non-Primitive data types
numbers=[1,2,3,4]
coordinates=(10.0,30.0)
names={"Prithi":"Thiru"}
a={1,2,3,4,}

print(type(numbers))
print(type(coordinates))
print(type(names))
print(type(a))
```
